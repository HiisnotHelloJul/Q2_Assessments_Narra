| Operation | Function Name | Parameters Needed | Process Inside the Function | Return Value |
| :--- | :--- | :--- | :--- | :--- |
| Addition | add_numbers | num1, num2 | Adds the two numbers num1 + num2 | The sum of the two numbers |
| Subtraction | subtract_numbers | num1, num2 | Subtracts the two numbers num1 - num2 | The difference between the two numbers |
| Multiplication | multiply_numbers | num1, num2 | Multiplies the two numbers num1 * num2 | The product of the two numbers |
| Division | divide_numbers | num1, num2 | Divides the two numbers num1 / num2 | The quotient of the two numbers |

```python
# Ask the user to input the numbers
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Operations
def add_numbers(num1, num2):
  sum = num1 + num2
  return sum

def subtract_numbers(num1, num2):
  difference = num1 - num2
  return difference

def multiply_numbers(num1, num2):
  product = num1 * num2
  return product

def divide_numbers(num1, num2):
  if num2 == 0:
    return "ERROR! Cannot divide by zero."
    return num1 / num2

# Ask the user to select an operation
operation = float(input("Select the operation (1-Addition, 2-Subtraction, 3-Multipliaction, 4-Division)"))

# Choices of the operations
if operation == 1:
  result = add_numbers(num1, num2)
  print("The sum of the numbers is ", result)

elif operation == 2:
  result = subtract_numbers(num1, num2)
  print("The difference between the numbers is ", result)

elif operation == 3:
  result = multiply_numbers(num1, num2)
  print("The product of the numbers is ", result)

elif operation == 4:
  result = divide_numbers(num1, num2)
  print("The quotient of the numbers is ", result)

else:
  print("Please enter a valid number.")
```
